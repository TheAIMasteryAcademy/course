# Lesson 0.2: Explore the penguins

This notebook answers the five dataset questions from the lesson, in a few lines of Python. Run each cell from top to bottom with **Shift + Enter**.

```python
from pathlib import Path
import pandas as pd

# Find the data file, wherever this notebook is opened from
here = Path.cwd()
csv = next(p / "data/penguins/penguins.csv" for p in [here, *here.parents] if (p / "data/penguins/penguins.csv").exists())
df = pd.read_csv(csv)
print("Loaded", len(df), "rows")
```

## 1. What is one row?

Look at the first five rows. Each row is one penguin.

```python
df.head()
```

## 2. What's the shape?

Rows first, then columns.

```python
df.shape
```

## 3. What type is each column?

`float64` means numbers with decimals. Text labels show as `object` or `str`, depending on your pandas version.

```python
df.dtypes
```

```python
numerical = df.select_dtypes("number").columns.tolist()
categorical = df.select_dtypes(exclude="number").columns.tolist()
print("Numerical:", numerical)
print("Categorical:", categorical)
```

You average numbers and you count categories.

```python
print("Average body mass:", round(df["body_mass_g"].mean() / 1000, 1), "kg")
df["species"].value_counts()
```

## 4. What's missing or strange?

Count the empty cells in each column.

```python
df.isna().sum()
```

These two penguins have no measurements at all:

```python
df[df["body_mass_g"].isna()]
```
